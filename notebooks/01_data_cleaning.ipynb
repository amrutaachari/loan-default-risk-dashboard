```python
# Work on a copy, so the original df stays untouched
df_clean = df.copy()
```

```python
# 1. Remove duplicate rows
df_clean = df_clean.drop_duplicates()
```

```python
# 2. Remove impossible values
df_clean = df_clean[df_clean["person_age"] <= 100]
df_clean = df_clean[df_clean["person_emp_length"].isna() | (df_clean["person_emp_length"] <= 60)]
```

```python
# 3. Fill missing values with the median
df_clean["person_emp_length"] = df_clean["person_emp_length"].fillna(df_clean["person_emp_length"].median())
df_clean["loan_int_rate"] = df_clean["loan_int_rate"].fillna(df_clean["loan_int_rate"].median())
```

```python
# 4. Check the result
print(df_clean.shape)
print(df_clean.isnull().sum())
```

```python
# 5. Save the cleaned file
df_clean.to_csv("data/processed/loan_data_clean.csv", index=False)
```
